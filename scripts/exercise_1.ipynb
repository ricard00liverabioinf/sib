```python
import numpy as np
from si.io.csv_file import read_csv

# 1.1) Carregar o dataset iris.csv
iris_path = 'datasets/iris.csv' # Garante que tens o iris completo ou ajusta o caminho
dataset = read_csv(iris_path, sep=',', features=True, label=True)

# 1.2) Selecionar a penúltima variável independente
penultimate_feature = dataset.X[:, -2]
print("1.2) Dimensão do array resultante:", penultimate_feature.shape)

# 1.3) Média das últimas 10 amostras para cada feature
last_10_samples = dataset.X[-10:, :]
mean_last_10 = np.mean(last_10_samples, axis=0)
print("1.3) Média das últimas 10 amostras por feature:", mean_last_10)

# 1.4) Selecionar amostras onde todas as features são <= 6
mask_lte_6 = np.all(dataset.X <= 6, axis=1)
samples_lte_6 = dataset.X[mask_lte_6]
print("1.4) Número de amostras com todas as features <= 6:", len(samples_lte_6))

# 1.5) Selecionar amostras com classe diferente de 'Iris-setosa'
mask_not_setosa = dataset.y != 'Iris-setosa'
samples_not_setosa = dataset.X[mask_not_setosa]
print("1.5) Número de amostras com classe diferente de 'Iris-setosa':", len(samples_not_setosa))
```
